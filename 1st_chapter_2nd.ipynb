```python
#adding digits  in a number
sum =0
num= int(input("enter a num"))
while(num!=0):
    rem=num%10
    sum=sum+rem
    num=num//10
    print(sum)
```

```python
# reversing a number
sum =0
num= int(input("enter a num"))
while(num!=0):
    rem=num%10
    sum=sum*10+rem
    num=num//10
print(sum)
```

```python
# prime num or not
n=int(input("enter a number to check"))
for i in range(2,n):
    if(n%i==0):
        print("its not a prime")
        break
else:
    print(n,"its a prime number")
```
