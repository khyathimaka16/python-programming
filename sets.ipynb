```python
#write a python program that prints missing numbers betwen 1 to n
n = int(input())
input_numbers_str = input().split(" ")

# Process input numbers, filtering out any empty strings from split
present_numbers_list = [int(i) for i in input_numbers_str if i.strip()]
present_numbers_set = __builtins__.set(present_numbers_list)

# Initialize a set to store missing numbers
missing_numbers_result_set = __builtins__.set()

# Iterate from 1 to n and add numbers not found in present_numbers_set
for num in range(1, n + 1):
    if num not in present_numbers_set:
        missing_numbers_result_set.add(num)

# Print the set of missing numbers
print(missing_numbers_result_set)
```

```python
#set us another built in datastructure
#it is a collection of heetrogenious elements
#it is unorderd and does't allow duplicate
#ther set is mutable but the set elements are immutable(hashable)
#as it is unorderd indexing is not possible
#SYNTAX of creating
        # s=set()
        #s={1,2,35,"hello"}
#set operations
  #union(|)
  #intersection(&)
  #diffrence(-)
  #symmetric diffrence(^)
  # Define two sample sets
set_A = {1, 2, 3, 4}
set_B = {3, 4, 5, 6}

print(f"Set A: {set_A}")
print(f"Set B: {set_B}\n")


op = set_A | set_B
method = set_A.union(set_B)
print(f"Union (A ∪ B): {op}")


op = set_A & set_B
method = set_A.intersection(set_B)
print(f"Intersection (A ∩ B): {op}")


op = set_A - set_B
method = set_A.difference(set_B)
print(f"Difference (A - B): {op}")


op = set_A ^ set_B
method = set_A.symmetric_difference(set_B)
print(f"Symmetric Difference (A Δ B): {op}")
```

```python
#set methods
    #add(add single element)
    #update(add multiple)
    #remove
    #discard
    #pop
    #clear
    #union
    #intersection
    #differance
    #symmetric difference
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))    # True
print(B.issuperset(A))  # True
print(A.isdisjoint(B))  # False (they share 1 and 2)
```

```python
# Demonstration of various set methods in Python

# 1. Initialize a sample set
s = {1, 2, 3}
print(f"Initial set s: {s}")

# 2. add(element) - Adds a single element
s.add(4)
print(f"After add(4): {s}")

# 3. update(iterable) - Adds multiple elements
s.update([5, 6, 7])
print(f"After update([5, 6, 7]): {s}")

# 4. remove(element) - Removes an element; raises KeyError if not present
s.remove(5)
print(f"After remove(5): {s}")

# 5. discard(element) - Removes an element; does NOT raise an error if absent
s.discard(10)  # 10 is not in the set, but no error is raised
s.discard(6)
print(f"After discard(6): {s}")

# 6. pop() - Removes and returns an arbitrary element
popped_val = s.pop()
print(f"Popped element: {popped_val}")
print(f"After pop(): {s}")

# 7. clear() - Removes all elements from the set
temp_set = {1, 2, 3}
temp_set.clear()
print(f"After clear() on temp_set: {temp_set}")

# --- Set Operations with another set ---
set_A = {1, 2, 3, 4}
set_B = {3, 4, 5, 6}
print(f"\nNew Set A: {s   et_A}")
print(f"New Set B: {set_B}")

# 8. union() - All unique elements from both sets
print(f"Union (A | B): {set_A.union(set_B)}")

# 9. intersection() - Elements present in both sets
print(f"Intersection (A & B): {set_A.intersection(set_B)}")

# 10. difference() - Elements in A but not in B
print(f"Difference (A - B): {set_A.difference(set_B)}")

# 11. symmetric_difference() - Elements in either A or B, but not both
print(f"Symmetric Difference (A ^ B): {set_A.symmetric_difference(set_B)}")
```
