```python
#LIST BUILT IN FUNCTIONS
#1.len()
#2.sum()
#3.min()
#4.max()
#5.reversed()
#6.sorted()
L6=[6,2,9,4,5,1,7,8,3,10]
print("sum",sum(L6))
print("length",len(L6))
print("min",min(L6))
print("max",max(L6))
print("sorted",sorted(L6))
print("reversed",list(reversed(L6)))
```

```python
#append
course=["python","java","dsa","math"]
course.append("java script")
print(course)
```

```python
#extend
branch=["aids","cse","ece"]
branch.extend(course)
print(branch)
```

```python
#insert
branch.insert(0,"civil")
print(branch)
```

```python
#pop
branch.pop(0)
print(branch)
```
