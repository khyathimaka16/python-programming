```python
#tuple is a collection of hetrogeous elements
#it is orederd and allows duplicates
#tupple is immmutable
#SYNTAX t1=() empty tupple
#t1=(1,23,34,"hello")
#__________NOTE_________
#single=(5,)#this is tuple
#t=(5)#this is just an integer
#size of the tuple is less than the size of the list
#tuple operations
t1=(1,2,3,45,"hello")
t2=(12,3,4,5,6,7,8)
print(t1+t2)#concationation
print(t1*2)#repetation
print(t1[2])#indexing
print(t2[0:4])#slicing
```

```python
#the only to method we use for tuple are
#count and index
#the fuunctions are
#len()
#max()
#sum()
#sorted()
t3=(1,2,3,4,5,6,7,78,6,5)
t4=(3,4,5,56,6,778,88, 8,2)
print(len(t3))
print(max(t4))
print(sum(t3))
print(min(t4))
print(sorted(t3))
```
